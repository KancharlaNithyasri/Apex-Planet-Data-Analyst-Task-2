```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv(r"C:\Users\nithish\OneDrive\Desktop\Sample - Superstore.csv" , encoding='latin1')

print(df.head())
```

```python
print(df.shape)
```

```python
#DESCRIPTIVE STATISTICS
print(df.describe())
```

```python
print(df.info())
```

```python
#HISTOGRAM
plt.figure(figsize=(8,4))
sns.histplot(df['Quantity'],kde=True,color='blue' , bins=14)
plt.title("Distribution of Order Quantity")
plt.xlabel("Quantity")
plt.ylabel('Count')
plt.show()
```

```python
#CATEGORY
print(df['Category'].value_counts())
```

```python
print(df['Category'].value_counts(normalize=True)*100)
```

```python
#BAR CHART
plt.figure(figsize=(10,5))
sns.countplot(data=df,x='Sub-Category',order=df['Sub-Category'].value_counts().index,hue='Sub-Category',
    legend=False,palette='viridis')
plt.title('Number of Orders by Product Sub-Category')
plt.xticks(rotation=45)
plt.ylabel('Number of Orders')
plt.show()
```
